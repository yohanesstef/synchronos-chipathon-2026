```python
from pygmid import Lookup as lk
import numpy as np
import scipy.constants as sc

# read data for GF180MCU NMOS and PMOS device
# the range of channel lengths is 0.28 to 3
# the range for VGS, VDS is 0 to 3.3
n = lk('../../bmurmann_gmoverid/gf180mcuD/simulation/nfet_03v3.mat')
p = lk('../../bmurmann_gmoverid/gf180mcuD/simulation/pfet_03v3.mat')
```

```python
#VCO Specification
Vsup = 3.3 #Supply voltage = 3.3V
Fmax = 400e6 #Maximum output frequency = 400MHz > 350MHz for PVT variation
Vswing = 1.2 #Output voltage swing = 1.2V
Vocm = 2.7 #Output common-mode voltage = 2.7V
Itail = 240e-6 #Total tail current = 240uA

#PMOS Symmetric Load
Id = Itail/2
gm_1 = Itail/Vswing
gmid_1 = gm_1/Id
print(gmid_1)
Length_1 = 0.28 #L = 0.28um
Vgs_1 = 1.2
jd_1 = p.lookup('ID_W', GM_ID=gmid_1, L=Length_1)
#vgs_1 = p.lookup('VT', GM_ID=gmid_1, L=Length_1)
#print(vgs_1)
print(jd_1)
w_1 = Id/jd_1
print(w_1)
```
