```python
import numpy as np
import pandas as pd
from pygmid import Lookup as lk
```

#### Technology data

```python
n = lk('../gf180mcuD/simulation/nfet_03v3.mat')
p = lk('../gf180mcuD/simulation/pfet_03v3.mat')
```

```python
id = 100e-6

l = 0.6
gm_id = np.array([5, 10, 20])

gm = id * gm_id
jd = n.lookup('ID_W', GM_ID=gm_id, L=l)
w = id/jd

cgg_w = n.lookup('CGG_W', GM_ID=gm_id, L=l)
cgg = w*cgg_w
ft = gm/cgg/2/np.pi

df = pd.DataFrame({
    'gm_id': gm_id,
    'id': np.full(len(gm_id), id),
    'jd': jd,
    'w': w,
    'cgg': cgg,
    'ft': ft
})

print(df)
```
