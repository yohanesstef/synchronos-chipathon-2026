```python
import numpy as np
import pandas as pd
from pygmid import Lookup as lk
```

#### Technology data

```python
n = lk('../gf180mcuD/simulation/nfet_03v3.mat')
```

#### Specifications

```python
gm = 100e-6
```

#### Design choices

```python
l = 0.3
gm_id = np.array([5, 10, 20])
```

#### Sizing and benchmarking

```python
id = gm/gm_id
jd = n.lookup('ID_W', GM_ID=gm_id, L=l)
w = id/jd

cgg_w = n.lookup('CGG_W', GM_ID=gm_id, L=l)
cgg = w*cgg_w
ft = gm/cgg/2/np.pi

df = pd.DataFrame({
    'gm_id': gm_id,
    'id': id,
    'jd': jd,
    'w': w,
    'cgg': cgg,
    'ft': ft
})

print(df)
```
