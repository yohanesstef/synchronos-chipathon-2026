```python
import matplotlib.pyplot as plt
import numpy as np

# Data extracted from the provided image
vctrl = np.array([0.35, 0.825, 1.2375, 1.65, 2.0625, 2.475, 2.6, 2.75])
freq_hz = np.array([3.88E+08, 2.77E+08, 1.99E+08, 1.26E+08, 5.37E+07, 3.26E+06, 3.94E+05, 1.84E+04])

# Convert frequency to MHz for cleaner axis labels
freq_mhz = freq_hz / 1e6

# Calculate KVCO (Derivative of Frequency w.r.t Vctrl)
# Note: Using absolute value because frequency decreases as Vctrl increases
kvco_mhz_per_v = np.abs(np.gradient(freq_mhz, vctrl))

# Create a figure with two subplots side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# --- Plot 1: VCO Tuning Curve (Log Scale) ---
ax1.semilogy(vctrl, freq_hz, marker='o', markersize=8, linestyle='-', linewidth=2, color='#0052cc')
ax1.set_title('VCO Tuning Curve', fontsize=16, fontweight='bold')
ax1.set_xlabel('Control Voltage, $V_{ctrl}$ (V)', fontsize=14)
ax1.set_ylabel('Frequency (Hz)', fontsize=14)
ax1.grid(True, which="both", ls="--", alpha=0.6)
ax1.tick_params(axis='both', labelsize=12)

# --- Plot 2: KVCO (Gain) Plot ---
ax2.plot(vctrl, kvco_mhz_per_v, marker='s', markersize=8, linestyle='-', linewidth=2, color='#cc0000')
ax2.set_title('VCO Gain ($K_{VCO}$)', fontsize=16, fontweight='bold')
ax2.set_xlabel('Control Voltage, $V_{ctrl}$ (V)', fontsize=14)
ax2.set_ylabel('$K_{VCO}$ (MHz / V)', fontsize=14)
ax2.grid(True, ls="--", alpha=0.6)
ax2.tick_params(axis='both', labelsize=12)

# Adjust layout and show/save
plt.tight_layout()
plt.savefig('vco_performance_plots.png', dpi=300)
plt.show()
```
